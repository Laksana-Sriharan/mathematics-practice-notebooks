```python
import numpy as np

# Given result vector
R = np.array([20, 10, -25])

# Since 2u + 3v + 4w = R and v=2u, w=3u
# => 20u = R
u = R / 20
v = 2 * u
w = 3 * u

print("u =", u)
print("v =", v)
print("w =", w)
```

```python
import numpy as np

u = np.array([2, -1, 2])
v = np.array([5, 0, 2])
x = np.array([-7, 1, 2])
y = np.array([0, 8, 3])

# a
a = np.dot(u + v, x + y)

# b
b = np.dot((np.dot(3*u, x)) * v, y)

# c
c = np.dot(u, x) - 4 * np.dot(v, y)

# d
d = np.dot(u, x) + 3*np.dot(u, y) - np.dot(v, y)

# e
e = np.dot(2 * (v + u), u) - 5 * np.dot(u, y)

# f
f = np.dot(4 * u, x) + 6 * np.dot(v, (3*x - y))

print("a", a, sep =": ")
print("b", b, sep =": ")
print("c", c, sep =": ")
print("d", d, sep =": ")
print("e", e, sep =": ")
print("f", f, sep =": ")
```

```python
import numpy as np

a = np.array([-3, 5]),
b = np.array([1, 0, -7]),
c = np.array([2, 2, 2, 2]),
d = np.array([1, -1, 2, -2, 3]),
e = np.array([1, 2, 3, 4, 5, 6])

print("a)", "||u|| =", np.linalg.norm(a))
print("b)", "||u|| =", np.linalg.norm(b))
print("c)", "||u|| =", np.linalg.norm(c))
print("d)", "||u|| =", np.linalg.norm(d))
print("e)", "||u|| =", np.linalg.norm(e))
```

```python
import numpy as np

def angle_between(u, v):
    cos_theta = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))
    angle = np.degrees(np.arccos(cos_theta))
    return angle

# a
u = np.array([3, 4])
v = np.array([5, 12])
print("a) angle =", angle_between(u, v))

# b
u = np.array([4, 6])
v = np.array([-3, 2])
print("b) angle =", angle_between(u, v))

# c
u = np.array([2, 2, -1])
v = np.array([5, -4, 2])
print("c) angle =", angle_between(u, v))
```

```python
import sympy as sp

b = sp.symbols('b')
# (12 + b) / (3 * sqrt(50 + b^2)) = 1/3

expr = (12 + b) / (3 * sp.sqrt(50 + b**2)) - sp.Rational(1, 3)

solution = sp.solve(expr, b)

print("b =", solution)
```

```python
import numpy as np
import sympy as sp

a = sp.symbols('a')
expr = (-6*a) + (a**3) + (2*a)
solutions = sp.solve(expr, a)
print("Values of a:", solutions)
```

```python
u = np.array([2, -3, 1])
v = np.array([7, -5, 1])

# a)
unit_u = u / np.linalg.norm(u)
print("Unit vector of u", unit_u)

# b) 
cross = np.cross(u, v)
unit_perp = cross / np.linalg.norm(cross)
print("Unit vector perpendicular to both u and v:", unit_perp)
```
