```python
import numpy as np
A = np.array([[3,1],[0,2],[-6,1]])
B = np.array([[2,0,1],[2,1,1]])

print(np.matmul(A,B))
print(np.matmul(B,A))
```

```python
import numpy as np
A = np.array([[1,3,2]])
B = np.array([[1,-1,0],[0,2,4],[2,-2,0]])

print(np.matmul(A,B))
#print(np.matmul(B,A)) Multiplication BA is not defined
```

```python
#Transpose of a matrix
A = np.array([[1,2,-1],[0,4,1]])
A.transpose()
```

```python
print(A.transpose().transpose())
```

```python
A = np.array([[1,2,-1],[0,4,1],[2,1,3]])
B = np.array([[4,3,-2],[2,7,3],[5,1,2]])
print(A)
print(B)
```

```python
print((A+B).transpose())
```

```python
print(A.transpose()+B.transpose())
```

```python
print((2*A).transpose())
print((A.transpose())*2)
```

```python
print(np.matmul(A,B).transpose())
print(np.matmul(B.transpose(),A.transpose()))
```
