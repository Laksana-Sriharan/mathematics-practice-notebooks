```python
import numpy as np
y = np.array([5000,6000,7000,8000,9000])
g = 2*y-8000
p = np.array([0.2,0.3,0.2,0.2,0.1])
print(np.dot(g,p))
```

```python
y=np.array([0,1,2,3])
p=np.array([1/8,1/4,3/8,1/4])
print("Mean", np.dot(y,p))
```

```python
v=(y-1.75)**2
print("Var:",np.dot(v,p))
```

```python
import math
print("Std Dev:",round(math.sqrt(np.dot(v,p)),4))
```
