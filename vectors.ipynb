```python
# Norm
import numpy as np

v = np.array([-2,-3])
norm = np.linalg.norm(v)
print(round(norm,3))

norm1 = np.sum(v**2)
print(norm1)
```

```python
import numpy as np

w = np.array([1,-2,3])
norm = np.linalg.norm(w)
print(round(norm,3))

norm1 = np.sum(w**2)
print(norm1)
```

```python
# Dot Product
import numpy as np

x = np.array([2,3])
y = np.array([4,2])

result = np.dot(x,y)
print(result)
```
