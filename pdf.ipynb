```python
from scipy.stats import uniform

print(round(uniform.cdf(30, loc = 0, scale = 30 )-uniform.cdf(25, loc = 0, scale = 30 ),4))
```

```python
from scipy.stats import expon
beta = 2.4

print(round(1-expon.cdf(3, scale = beta ),4))
```

```python
from scipy.stats import expon
beta = 2.4

print(round(expon.cdf(3, scale = beta )-expon.cdf(2, scale = beta ),4))
```

```python
import math

sample_mean = 33
sample_var = 256
sample_n = 64

CL = 0.95
z_val = norm.ppf(0.95)

LCL = sample_mean - z_val*(math.sqrt(sample_var/sample_n))
UCL = sample_mean + z_val*(math.sqrt(sample_var/sample_n))
print(f"Confidence Interval: ({LCL:.3F},{UCL:.3f})")
```
