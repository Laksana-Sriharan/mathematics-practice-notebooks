```python
import numpy as np

def bayesTheorem(p, PX):
    return print(p*PX/sum(p*PX))
```

```python
p = [0.001,0.999]
px = [0.92,0.04] 
p = np.array(p)
px = np.array(px)

bayesTheorem(p, px)
```
