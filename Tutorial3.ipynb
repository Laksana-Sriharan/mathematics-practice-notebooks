```python
from sympy import Matrix

A = Matrix([[1,2,3], [4,-2,3], [0,5,-1]])
print(A.det())
```

```python
from sympy import Matrix

A = Matrix([[4,-1,-2], [0,2,-3], [5,2,1]])
print(A.det())
```

```python
from sympy import Matrix, symbols

a,b = symbols('a b')
A = Matrix([[a-b, b], [b, a+b]])
print(A.det())
```

```python
from sympy import Matrix

A = Matrix([[1,-1,2], [2,-1,1], [1,1,-1]])
print(A.inv())
```

```python
from sympy import Matrix

A = Matrix([[1, 2, -4], [-1, -1, 5], [2, 7, -3]])
print(A.inv())
```

```python
from sympy import Matrix, symbols

k = symbols('k')
A = Matrix([[3, 2, 4], [2, 0, 2], [4, 2, k]])
print("Determinant: ", A.det())
print("Inverse: ",A.inv())
```
