```python
from sympy import symbols, diff
x,y = symbols('x y')
f = 2*x + 5*y - 3
diff(f,x)
```

```python
diff(f,y)
```

```python
from sympy import simplify
z = (x**2 + y**2)/(x+y)
simplify(diff(z,x) - diff(z,y))**2 #LHS
```

```python
#RHS
simplify(4*(1-diff(z,x) - diff(z,y)))
```

```python
lhs = simplify(diff(z,x) - diff(z,y))**2
rhs = simplify(4*(1-diff(z,x) - diff(z,y)))
lhs.equals(rhs)
```

```python
from sympy import log
u = log((x**2 + y**2)/(x*y))
ex1 = diff(u,x,y)
ex2 = diff(u,y,x)
ex1.equals(ex2)
```
