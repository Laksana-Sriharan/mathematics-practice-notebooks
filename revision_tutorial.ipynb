```python
import numpy as np
a = np.array([2,2,-2])
b = np.array([6,-3,2])
print(np.dot(a,b))
```

```python
#Scalar projection
from scipy.linalg import norm
print(np.dot(a,b)/norm(a))
```

```python
#Vector Projection
print(np.dot(a,b)/(norm(a)*norm(a))*2)
```
