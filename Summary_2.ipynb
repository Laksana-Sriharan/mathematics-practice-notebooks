```python
#Adjoint of a matrix
import sympy as sp
from sympy import Matrix
A = Matrix([[2,1,-1],[3,1,-2],[1,0,1]])
A.adjugate()
```

```python
print("Determinant of A = ", A.det())
print("Inverse of A = \n", (A.adjugate()/A.det()))
```

```python
#Inverse of a Matrix
A.inv()
```

```python
#Eigenvalues
import numpy as np
A = np.array([[1,3],[2,-4]])
w,v = np.linalg.eig(A)
print("Eigenvalue: ", w)
print("Eigenvector: ", v)
```

```python
import numpy as np
B = np.array([[8,-6,2],[-6,7,-4],[2,-4,3]])
w,v = np.linalg.eig(B)
print("Eigenvalue: ", w)
print("Eigenvector: ", v)
```

```python
#multivariate functions
from sympy import symbols, diff, sin, cos
x,y,t = symbols('x,y,t')
xt = sin(2*t)
yt = cos(t)
f = x**2*y + 3*x*y**4
res = diff(f,x)*diff(xt,t) + diff(f,y)*diff(yt,t)
res
```
