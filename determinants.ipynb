```python
from sympy import Matrix

A = Matrix([[4, 2], [1, 3]])
print(A.det())
```

```python
from sympy import Matrix

A = Matrix([[1,5,0], [2,4,-1],[0,-2,0]])
print(A.det())
```

```python
A = Matrix([[3,-7,8,9,-6], [0,2,-5,7,3],[0,0,1,5,0],[0,0,2,4,-1],[0,0,0,-2,0]])
print(A.det())
```

```python
#Verify Multiplicative property
A = Matrix([[6,1],[3,2]])
B = Matrix([[4,3],[1,2]])
print(A)
print(B)

print(A.det())
print(B.det())
print(A.det()*B.det())
print((A*B).det())
```
