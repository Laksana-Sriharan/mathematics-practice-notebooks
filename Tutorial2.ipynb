```python
from sympy import symbols, Eq, solve
#Define th e symbols
a,b = symbols('a,b')

eq1 = Eq((3*a-6*b),0)
print(eq1)
eq2 = Eq((2*a-4*b),0)
print(eq2)

print(solve((eq1,eq2), (a,b)))
```

```python
import matplotlib.pyplot as plt

plt.quiver(0, 0, 4, 2, angles='xy', scale_units='xy', scale=1, color='blue', label='c1')
plt.quiver(0, 0, -8, -4, angles='xy', scale_units='xy', scale=1, color='red', label='c2')

plt.xlim(-10, 6)
plt.ylim(-6, 6)

plt.axhline(0, color='black')
plt.axvline(0, color='black')

plt.grid()
plt.legend()
plt.title("Vectors c1 and c2")
plt.show()
```
